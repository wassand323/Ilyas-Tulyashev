# Базовый EDA датасета фильмов



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)

df = pd.read_csv("film.csv", sep=";", encoding="cp1252", skiprows=[1])

df.columns = df.columns.str.strip()

df.head()
```

## 1. Размер и структура данных

```python
print(f"Количество строк: {df.shape[0]}")
print(f"Количество столбцов: {df.shape[1]}")
display(df.info())
display(df.describe(include="all").T)
```

## 2. Пропущенные значения

```python
missing = pd.DataFrame({
    "Пропусков": df.isna().sum(),
    "Доля пропусков, %": (df.isna().mean() * 100).round(2)
}).sort_values("Пропусков", ascending=False)

display(missing[missing["Пропусков"] > 0])

missing[missing["Пропусков"] > 0]["Пропусков"].sort_values().plot(
    kind="barh", title="Количество пропусков по столбцам"
)
plt.xlabel("Количество пропусков")
plt.tight_layout()
plt.show()
```

## 3. Дубликаты

```python
print("Полных дубликатов:", df.duplicated().sum())

print("Повторяющихся названий фильмов:", df["Title"].duplicated().sum())
```

## 4. Числовые признаки: год, длительность и популярность

```python
numeric_cols = ["Year", "Length", "Popularity"]
display(df[numeric_cols].describe().T)

fig, axes = plt.subplots(1, 3, figsize=(16, 4))
for col, ax in zip(numeric_cols, axes):
    sns.histplot(df[col].dropna(), kde=True, ax=ax)
    ax.set_title(f"Распределение: {col}")
plt.tight_layout()
plt.show()
```

## 5. Количество фильмов по жанрам

```python
genre_counts = df["Subject"].value_counts()
display(genre_counts.to_frame("Количество фильмов"))

genre_counts.plot(kind="bar", color="steelblue")
plt.title("Количество фильмов по жанрам")
plt.xlabel("Жанр")
plt.ylabel("Количество фильмов")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## 6. Динамика количества фильмов по годам

```python
films_by_year = df["Year"].value_counts().sort_index()
films_by_year.plot()
plt.title("Количество фильмов по годам")
plt.xlabel("Год")
plt.ylabel("Количество фильмов")
plt.tight_layout()
plt.show()

df["Decade"] = (df["Year"] // 10) * 10
display(df["Decade"].value_counts().sort_index().to_frame("Количество фильмов"))
```

## 7. Наличие наград

```python
awards_counts = df["Awards"].value_counts(dropna=False)
display(awards_counts.to_frame("Количество"))

awards_counts.plot(kind="bar", color=["slateblue", "darkorange"])
plt.title("Фильмы с отметкой о наградах")
plt.xlabel("Awards")
plt.ylabel("Количество фильмов")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 8. Часто встречающиеся режиссёры и актёры

```python
print("Топ-10 режиссёров по количеству фильмов:")
display(df["Director"].value_counts().head(10).to_frame("Количество фильмов"))

print("Топ-10 актёров по количеству фильмов:")
display(df["Actor"].value_counts().head(10).to_frame("Количество фильмов"))

print("Топ-10 актрис по количеству фильмов:")
display(df["Actress"].value_counts().head(10).to_frame("Количество фильмов"))
```

## 9. Корреляция числовых признаков

```python
corr = df[["Year", "Length", "Popularity"]].corr()
display(corr)

sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Корреляция числовых признаков")
plt.tight_layout()
plt.show()
```
