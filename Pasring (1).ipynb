```python
import requests
import csv
import pandas as pd
```

```python
API_KEY = "2ZM0XVY-AWW48DN-Q6THJ1R-VRPYDS1"

def fetch_top_250():
    print("Подключаемся и скачиваем")
    url = "https://api.kinopoisk.dev/v1.4/movie?page=1&limit=250&lists=top250"
    
    headers = {
        "accept": "application/json",
        "X-API-KEY": API_KEY
    }
    
    try:
        response = requests.get(url, headers=headers)
        response.raise_for_status()
        
        data = response.json()
        movies = data.get('docs', [])
        
        print(f"Успешно скачано фильмов: {len(movies)}")
        return movies
    except Exception as e:
        print(f"Ошибка при скачивании: {e}")
        return []

movies_data = fetch_top_250()
```

```python
def save_to_csv(movies_list, filename="top_250_movies.csv"):
    fieldnames = ['Название', 'Год', 'Страна', 'Описание', 'Жанр', 'Рейтинг']
    
    with open(filename, mode='w', encoding='utf-8-sig', newline='') as file:
        writer = csv.DictWriter(file, fieldnames=fieldnames, delimiter=';')
        writer.writeheader()
        
        for movie in movies_list:
            name = movie.get("name") or "Без названия"
            year = movie.get("year", "")
            
            countries = ", ".join([c.get("name", "") for c in movie.get("countries", []) if c.get("name")])
            genres = ", ".join([g.get("name", "") for g in movie.get("genres", []) if g.get("name")])
            
            description = movie.get("description") or movie.get("shortDescription") or "Нет описания"
            description = description.replace("\n", " ").replace("\r", " ").strip()
            
            rating = movie.get("rating", {}).get("kp") or movie.get("rating", {}).get("imdb") or ""
            
            writer.writerow({
                'Название': name,
                'Год': year,
                'Страна': countries,
                'Описание': description,
                'Жанр': genres,
                'Рейтинг': rating
            })
            
    print(f"Готово! Успешно сохранено {len(movies_list)} фильмов в файл {filename}.")

if movies_data:
    save_to_csv(movies_data)
else:
    print("Cписок фильмов пуст.")
```
